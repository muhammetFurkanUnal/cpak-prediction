```python
# extract file names from folder
dataset_files_list = []
with open(dataset_files_list_path, "r") as f:
	dataset_files_list = f.readlines()  # length: 94

for i in range(len(dataset_files_list)):
	split_list = dataset_files_list[i].split(".")
	_id = split_list[0]
	l_or_r = split_list[1]
	dataset_files_list[i] = f"{_id}{l_or_r}"

dataset_files_list_unique = list(set(dataset_files_list)) 
print("number of samples: ", len(dataset_files_list_unique))
print(dataset_files_list_unique[0])
```
